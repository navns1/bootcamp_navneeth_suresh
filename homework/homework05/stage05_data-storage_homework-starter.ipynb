```python
# Load env variables
load_dotenv()
raw_dir = os.getenv("DATA_DIR_RAW")
processed_dir = os.getenv("DATA_DIR_PROCESSED")
```

```python
# Make sample DataFrame
data = {
    'toy_id': [1, 2, 3],
    'toy_name': ['Blocks', 'Train', 'Car'],
    'price': [19.99, 24.50, 9.99]
}
df = pd.DataFrame(data)
```

```python
# Create unique timestamp names
timestamp = datetime.now().strftime("%Y%m%d-%H%M")
csv_path = os.path.join(raw_dir, f"sample_{timestamp}.csv")
parquet_path = os.path.join(processed_dir, f"sample_{timestamp}.parquet")
```

```python
# Save the files using our tools
write_df(df, csv_path)
write_df(df, parquet_path)
```
